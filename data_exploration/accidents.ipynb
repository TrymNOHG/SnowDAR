## Analysis of Snow-related driving accidents in Norway

The data used for this exploration comes from Statens Vegvesen https://trine.atlas.vegvesen.no/.

```python
import pandas as pd
df = pd.read_csv("./TabellSVV.csv", delimiter=";")
```

```python
df.head()
```

```python
df.columns
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = df['År']  
deaths = df['Drept'] 
severe = df['Hardt skadd'] 
mild = df['Lettere skadd'] 
accidents = df['Sum'] 

plt.plot(x, deaths, label="Deaths")  
plt.plot(x, severe, label="Severe Accident")  
plt.plot(x, mild, label="Mild Accident")  
plt.plot(x, accidents, label="Total Accident")  
plt.xlabel("Year")  
plt.ylabel("Number of occurences")  
plt.title("Development of snow-related accidents in Norway")  
plt.legend()
plt.savefig("snow_accident_norway.png")
```
